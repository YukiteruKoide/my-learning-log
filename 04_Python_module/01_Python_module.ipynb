## Time

```python
import time
```

```python
print(time.time())
print(time.time()/(60*60*24*365))
```

```python
def fib(n):
  print(f'fib with {n} is runnninng...')
  if n < 2:
    return n
  else:
    return fib(n-1) + fib(n-2)

before = time.time()
fib(30)
after = time.time()
print(f'{after - before:.2f}')
```

```python
from functools import lru_cache

@lru_cache
def fib(n):
  print(f'fib with {n} is runnninng...')
  if n < 2:
    return n
  else:
    return fib(n-1) + fib(n-2)

before = time.time()
fib(30)
after = time.time()
print(f'{after - before:.2f}')
```

```python
import time

# ctime() 今のローカル時間を文字列で返す
print(time.ctime())

# .localtime() 今のローカル時間を構造化データで返す
localtime = time.localtime()
print(localtime)
print(localtime.tm_year)
print(localtime.tm_hour)
print(localtime.tm_min)
```

```python
# .sleep(secs) secs秒だけプログラムが待機する
sec = 10
print(f'{sec}秒待ってください')
time.sleep(sec)
print(f'{sec}秒経過しました！')
```

### Challenge 関数の実行時間を測るtimerデコレータを作ろう

```python
def timer(func):
  def inner(*args, **kwargs):
    before = time.time()
    func(*args, **kwargs)
    after = time.time()
    print(f"{func.__name__} took {after - before:.2f} seconds")
  return inner


@timer
def lazy_func(sec):
  print('I am working so hard...')
  time.sleep(sec)
  print('I am finally done!')

lazy_func(4)
```
