# 例外処理

```python
def split_bill(price):
  num = input('何人で割り勘しますか？')
  try:
    each = price / int(num)
  except ZeroDivisionError as e:
    print('0で割ることはできません。正しい値を入れてください')
    each = price
    print(e)
  except ValueError as e:
    print('数字を入れてください')
    each = price
  print(f'一人{each}円です')


if __name__ == '__main__':
  split_bill(10000)
```

## else と finally

```python
def split_bill(price):
  num = input('何人で割り勘しますか？')
  try:
    each = price / int(num)
  except ZeroDivisionError as e:
    print('0で割ることはできません。正しい値を入れてください')
    each = price
    print(e)
  except ValueError as e:
    print('数字を入れてください')
    each = price
  else:
    print(f'一人{each}円です')
  finally:
    print('ご利用ありがとうございます')


if __name__ == '__main__':
  split_bill(10000)
```

## 型のチェック

```python
class Animal:
  def __init__(self, name):
    self.name = name

  def walk(self):
    print(f'{self.name} is walking ')

class Dog(Animal):
  def bark(self):
    print(f'{self.name} is barking')

def walk_with_me(animal: Animal) -> None:
  #if isinstance(animal, Animal):
  method = getattr(animal, 'walk', None)
  if callable(method):
    animal.walk()
  else:
    raise TypeError('animal must be Animal type')


if __name__ == '__main__':
  pochi = Dog('Pochi')
  walk_with_me(pochi)
```

## tracebackモジュール

```python
import traceback

bill = {'burger' : 500, 'pasta' : 1400, 'soda' : 300}

def split_bill(price):
  num = input('何人で割り勘しますか？')
  try:
    each = price / int(num)
  except ZeroDivisionError as e:
    print('0で割ることはできません。正しい値を入れてください')
  except ValueError as e:
    print('数値を入力してください。')
  else:
    print(f'一人{each}円です')


def check(bill):
  total_bill = sum(bill.values())
  try:
    split_bill(total_bill)
  except:
    traceback.print_exc()
    print('エラーが出ました。')


if __name__ == '__main__':
  bill = {'burger' : 500, 'pasta' : 1400, 'soda' : 300}
  check(bill)
```
