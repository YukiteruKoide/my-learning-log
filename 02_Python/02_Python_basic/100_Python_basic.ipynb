## Challenge
ユーザに好きなフルーツを入力してもらい、そのフルーツがfruitsリストにあればそのフルーツを削除し、なければフルーツを追加する

```python
# リストの定義
fruits = ['apple', 'peach', 'melon', 'grapes']

# インプットの定義
fruit = input('好きなフルーツはありますか？')

# 条件分岐
if fruit in fruits:
  print(f'{fruit}が好きなんですね。差し上げます。')
  fruits.remove(fruit)
else :
  print(f'{fruit}が好きなんですね。仕入れました！')
  fruits.append(fruit)

# 結果の表示
print(f'今あるフルーツは{fruits}です')
```

## Challenge1
ユーザに好きなフルーツを入力してもらいfruitsリストの各フルーツに対して「好き、好きじゃない」をprint()してもらう

```python
# fruitsリストの定義
fruits = ['apple', 'peach', 'melon', 'grapes']

# inputしてもらう
favorite = input('好きなフルーツはありますか？')

# 判定を行う
for fruit in fruits:
  if favorite == fruit:
    print(f'{fruit}が好きなんですね！')
  else :
    print(f'{fruit}は好きじゃないんですね')
   
```

## Challenge2 
fruitsリストの各フルーツに対して「好き、好きじゃない」をユーザに聞いて、好きなフルーツリスト、好きじゃないフルーツリストを作成する

```python
# fruitsリストの定義
fruits = ['apple', 'peach', 'melon', 'grapes']
favorite_list = []
Not_favorite_list = []

# ユーザへのヒアリングループ
for fruit in fruits:
  favorite = input(f'{fruit}は好きですか？')
  if favorite == 'yes':
    favorite_list.append(fruit)
  else :
    Not_favorite_list.append(fruit)
print('好きなフルーツは'+str(favorite_list)+'です')
print('嫌いなフルーツは'+str(Not_favorite_list)+'です')
```

## FizzBuzzゲーム
1~ 50の数字をそれぞれprint()して、3の倍数でfizz, 5の倍数でbuzz、3と5の倍数でfizzbuzz


```python
for i in range(1, 51):
  # 15の倍数
  if i % 15 == 0:
    print('FizzBuzz')
  elif i % 3 == 0:
    print('Fizz')
  elif i % 5 == 0:
    print('Buzz')
  else:
    print(i)
```

## Challenge1
ユーザに年齢を聞き、18歳以上なら入れるカジノを作成する。カジノに入ったらゲームを選べるようにする。
- 1. ルーレット
- 2. ブラックジャック
- 3. ポーカー
上記のコードを1 ~ 3以外の文字が入力されたら再度ゲームを選べるようにする

```python
# 年齢を確認
age = input('年齢を教えてください')

# 年齢確認を実施
if int(age) < 18:
  print('18歳未満はお帰りください')
else:
  game = input('ゲームをお選びください')
  while True:
    if game == str(1):
      print('ルーレットですね')
      break
    elif game == str(2):
      print('ブラックジャックですね')
      break
    elif game == str(3):
      print('ポーカーですね')
      break
    else:
      print('そのゲームはありません')
      continue
```

## ループのelse

```python
# for else
fruits = ['apple', 'peach', 'melon', 'grapes']

for fruit in fruits:
  choice = input(f'あなたが探しているフルーツは{fruit}ですか？y/n')
  if choice == 'y':
    print('見つかって良かったですね')
    break
  else:
    print('そうですか…')
else:
  print('お探しのフルーツは見つかりませんでした')
```
