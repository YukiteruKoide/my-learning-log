# fileio

## open関数

```python
f = open('/Volumes/users/yukiteru_koide/learning/test.txt')
print(f)
```

```python
f = open('/Volumes/users/yukiteru_koide/learning/test.txt')
for line in f:
  print(line)
```

```python
f.close()
```

### with statement

```python
with open('/Volumes/users/yukiteru_koide/learning/test.txt') as f:
  for line in f:
    print(line, end="")
```

## read

```python
# .read()でファイルの中身を全てを１つのstringとして返す
with open('/Volumes/users/yukiteru_koide/learning/test.txt') as f:
  print(f.read())
```

## readline()

```python
# .readline()で１行ずつ読み込む
with open('/Volumes/users/yukiteru_koide/learning/test.txt') as f:
  line = f.readline()
  print(line)
```

```python
# .readline()で１行ずつ読み込む
with open('/Volumes/users/yukiteru_koide/learning/test.txt') as f:
  while line:
    print(line, end="")
    line = f.readline()
```

## .readlines()で全ての行をlistで返す

```python
# .readline()で１行ずつ読み込む
with open('/Volumes/users/yukiteru_koide/learning/test.txt') as f:
  line = f.readlines()
  print(line)
```
