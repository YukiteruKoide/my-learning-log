## read

```python
# .read()でファイルの中身を全てを１つのstringとして返す
with open('/Volumes/users/yukiteru_koide/learning/test.txt') as f:
  print(f.read())
```

## readline()

```python
# .readline()で１行ずつ読み込む
with open('/Volumes/users/yukiteru_koide/learning/test.txt') as f:
  line = f.readline()
  print(line)
```

```python
# .readline()で１行ずつ読み込む
with open('/Volumes/users/yukiteru_koide/learning/test.txt') as f:
  while line:
    print(line, end="")
    line = f.readline()
```

## .readlines()で全ての行をlistで返す

```python
# .readline()で１行ずつ読み込む
with open('/Volumes/users/yukiteru_koide/learning/test.txt') as f:
  line = f.readlines()
  print(line)
```

## ファイルの書き込み

### mode w

```python
with open('/Volumes/users/yukiteru_koide/learning/test.txt', mode='w') as f:
  f.write('test1\ntest2' )

# 上書き(truncate)されてしまうので注意が必要!!
```

```python
# .readline()で１行ずつ読み込む
with open('/Volumes/users/yukiteru_koide/learning/test.txt') as f:
  line = f.readlines()
  print(line)
```

```python
with open('/Volumes/users/yukiteru_koide/learning/test2.txt', mode='w') as f:
  f.write('new_test')

# 新しいファイルが作られる
```

### 様々なmode

```python
# mode = 'a'で追記モード
with open('/Volumes/users/yukiteru_koide/learning/test2.txt', mode='a') as f:
  f.write('new_test2')

# サポートされていない
```

```python
# mode = 'r'で読み書きモード　ただし上書きされる
with open('/Volumes/users/yukiteru_koide/learning/test2.txt', mode='r+') as f:
  f.write('new_test3')
```

```python
with open('/Volumes/users/yukiteru_koide/learning/test2.txt') as f:
  line = f.readlines()
  print(line)
```

```python
# mode = 'w+'で読み書きモード　ただしtruncateされる
with open('/Volumes/users/yukiteru_koide/learning/test2.txt', mode='w+') as f:
  f.write('new_test3')
```

```python
with open('/Volumes/users/yukiteru_koide/learning/test2.txt') as f:
  line = f.readlines()
  print(line)
```

## csvの取り扱い

```python
with open('/Volumes/users/yukiteru_koide/learning/example.csv') as f:
  print(f.read())
```

```python
import csv

# csv.reader()
with open('/Volumes/users/yukiteru_koide/learning/example.csv') as f:
  reader = csv.reader(f)
  for line in reader:
    print(line)
```

```python
with open('/Volumes/users/yukiteru_koide/learning/example.csv') as f:
  reader = csv.reader(f)
  for line in reader:
    print(line[1])
```

```python
# csv.DictReader()
with open('/Volumes/users/yukiteru_koide/learning/example.csv') as f:
  reader = csv.DictReader(f)
  for line in reader:
    print(line)
```

```python
# csv.DictReader()
with open('/Volumes/users/yukiteru_koide/learning/example.csv') as f:
  reader = csv.DictReader(f)
  for line in reader:
    print(line['Gender'])
```

```python
# csv.writer(f)
with open('/Volumes/users/yukiteru_koide/learning/sample.csv', mode='w') as f:
  writer = csv.writer(f)
  writer.writerow(['Name', 'Age'])
  writer.writerow(['Bob', 22])
  writer.writerow(['Alice', 25])
```
