## オブジェクト指向の基礎

```python
class Person(object):
  def __init__(self, name, age, gender):
    self.name = name
    self.age = age
    self.gender = gender

  def walk(self):
    print(f"{self.name} is walking")

  def run(self):
    print(f"{self.name} is running")


john = Person("John", "28", "male")
taro = Person("Taro", "18", "male")
emma = Person("Emma", 40, "female")

# インスタンス変数
print(john.name)
print(taro.age)
print(emma.gender)

# 変数の変更
print(john.age)
john.age = 29
print(john.age)

# メソッド
john.walk()
taro.walk()
emma.run()
```

### Challenge
- 銀行口座のAccountクラス
- 残高（balane）と口座名（name）をもとに口座（Account）を作る
- withdrawメソッドとdepositメソッドで残高を変更する
- 残高が足りなければ引き落とせないようにする
- 口座番号（account_number）は今まで作成された口座のIDとなるように連番を振る（0, 1, 2, ...）
- 残高が変更されたら口座名、口座番号とその残高を表示する

```python
class Account(object):

    count = 0  # Class variable

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
        self.account_number = Account.count
        Account.count += 1

    def withdraw(self, price):
        if price > self.balance:
            print('残高が足りません')
        else:
            self.balance -= price
        self.show_balance()

    def deposit(self, price):
        self.balance += price
        self.show_balance()

    def show_balance(self):
        print(f'{self.account_number}の残高は{self.balance}円です')


myaccount = Account(name='my name', balance=1000)
print(myaccount.name)
print(myaccount.balance)
myaccount.withdraw(200)
myaccount.deposit(500)

youraccount = Account(name='my name', balance=2000)
print(youraccount.account_number)
```

### Challege
- 前回のChallegeで作成したAccountクラスに、取引（transaction）を記録する仕組みを追加する
- 取引（transaction）として保持する情報は
  - withdraw/depositの金額
  - 新しい残高
  - 日時
- それぞれの取引情報をdictionaryとして保持し、それおをlistでインスタンス変数で保持すればOK
- 日時を作る関数をstaticmethodで作ってみよう

```python
import time

class Account(object):

    count = 0  # Class variable

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
        self.account_number = Account.count
        self.show_balance()
        self.transaction_history = []
        Account.count += 1

    def withdraw(self, price):
        if price > self.balance:
            print('残高が足りません')
        else:
            self.balance -= price
            self.show_balance()
            self.add_transaction(price)

    def deposit(self, price):
        self.balance += price
        self.show_balance()
        self.add_transaction(price)

    def show_balance(self):
        print(f'{self.account_number}の残高は{self.balance}円です')
      
    def add_transaction(self, price):
        transaction = {
          'withdraw/deposit': price,
          'new_balance': self.balance,
          'time': Account.get_time_str()
        }
        self.transaction_history.append(transaction)

    @staticmethod
    def get_time_str():
        current_time = time.localtime()
        return time.strftime('%Y-%m-%d %H:%M:%S', current_time)

# Create an instance of Account
myaccount = Account('John Doe', 1000)

myaccount.withdraw(200)
myaccount.deposit(500)
print(myaccount.transaction_history)
```

## クラスメソッド
引数にクラスを書く

```python
class MyClass:

  classmethodme = 0

  def mymethod(self):
    print(f'This is normal method!! from {self}')

  @staticmethod
  def mystaticmethod():
    print('This is static method!!')

  @classmethod
  def myclassmethod(cls):
    cls.classmethodme += 1
    print(f'This is class method!! and now the count is {cls.classmethodme}')

c = MyClass()
c.mymethod()
MyClass.mystaticmethod()
MyClass.myclassmethod()
MyClass.myclassmethod()
```

## private変数と名前修飾

```python
class Account:

  def __init__(self, balance):
    self._balance = balance

  def deposit(self, price):
    self._balance += price
    self.show_balance() 

  def withdraw(self, price):
    if self._balance < price:
      print('残高が足りません')
    else:
      self._balance -= price
    self.show_balance()
  
  def show_balance(self):
    print(f'残高は{self._balance}円です')
  
myaccount = Account(1000)
myaccount.deposit(500)
myaccount.withdraw(200)
```

## getterとsetter

```python
class Person:

  def __init__(self, name, age):
    self.name = name
    self._age = age
  
  def set_age(self, age):
    print('set_age called!!')
    if age < 0:
      print('0以上の値を入れてください')
    else:
      self._age = age

  def get_age(self):
    print('get_age called!!')
    return self._age

  age = property(get_age, set_age)  

john = Person("John", 28)
print(john.age)
```

## プロパティデコレータ

```python
class Person:

  def __init__(self, name, age):
    self.name = name
    self._age = age

  @property
  def age(self):
    print('get_age called!!')
    return self._age

  @age.setter
  def age(self, age):
    print('set_age called!!')
    if age < 0:
      print('0以上の値を入れてください')
    else:
      self._age = age

john = Person("John", 28)
print(john.age)
```

## オーバーライド

```python
class Car(object):
  def __init__(self, model_name, mileage, manufacturer):
    self.model_name = model_name
    self.mileage = mileage
    self.manufacturer = manufacturer

  def gas(self):
    print(f"{self.manufacturer}の{self.model_name}（燃費: {self.mileage})：アクセル全開!!")

  def brakes(self):
    print(f"{self.manufacturer}の{self.model_name}（燃費: {self.mileage})：ブレーキ!!")


class Truck(Car):
    def __init__(self, model_name, mileage, manufacturer, max_load, current_load):
      super().__init__(model_name, mileage, manufacturer)
      self._max_load = max_load
      self._loadings = 0

    def gas(self):
      if self._loadings > self._max_load:
        print(f'積載量が最大値を超えています')
        print(f'最大でも{self._loadings - self._max_load}tの荷物を降ろしてください')
      else:
        super().gas()

    def load(self, weight):
      if weight > 0:
        print(f'{weight}tの荷物を積みました')
        self._loadings += weight
      else:
        if self._loadings <= -weight:
          print(f'{self._loadings}t全ての荷物を荷物を降ろしました')
          self._loadings = 0
        else:
          self._loadings += weight
      print(f'現在の積載量は{self._loadings}tです')

isuzu_truck = Truck("トラックA", 20, "いすゞ", 10, 0)
isuzu_truck.gas()
isuzu_truck.load(5)
isuzu_truck.load(-3)
isuzu_truck.load(10)
isuzu_truck.gas()
isuzu_truck.load(-30)
```
