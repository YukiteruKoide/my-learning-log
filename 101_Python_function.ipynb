## *args, **kwargs

```python
# *args タプルでもらう
def get_average(*args):
  num = len(args)
  if num == 0:
    return 0
  total = sum(args)
  return total / num

average = get_average(1, 2, 3, 4, 5)
print(average)

# **kwargs　ディクショナリーでもらう
def kwargs_func(**kwargs):
  param1 = kwargs.get('param1')
  param2 = kwargs.get('param2')
  param3 = kwargs.get('param3')
  return param1 + param2 + param3

print(kwargs_func(param1=1, param2=2, param3=3))

# *と**はunpacking, operator
numbers = (1, 2, 3)
print(*numbers)

a = {'a': 1, 'b': 2}
b = {'c': 3, 'd': 4}
print({**a, **b})
```

## 参照渡し、値渡し

```python
# 参照渡し <-> 値渡し
def add_nums(a, b):
  print(f'第一引数aのIDは{id(a)}')
  print(f'第二引数bのIDは{id(b)}')
  return a + b

one = 1
two = 2
print(f'第一引数aのIDは{id(one)}')
print(f'第二引数bのIDは{id(two)}')
print(add_nums(one, two))
```

```python
def add_one(num):
  print(f'変更前のID:{id(num)}')
  num += 1
  print(f'変更後のID:{id(num)}')
  return num


one = 1
print(id(one))
print(f'関数呼び出し前のID:{id(one)}')
add_one(one)
print(f'関数呼び出し後のID:{id(one)}')
```

## type annotation

```python
# type annotation
def add_nums(num1: int, num2: int) -> int:
  return num1 + num2


# 動的型付け言語
one = 1
two = 2
print(add_nums(one, two))
```

## 変数のスコープ

```python
# 変数のスコープ
# ローカル変数
def print_name_local():
  first_name = 'John'
  last_name = 'Doe'
  print(f'私の名前は{first_name} {last_name}です')


print_name_local()

# グローバル変数(モジュール変数)
age = 30


def print_age():
  age = 20 # ローカル変数
  print(f'私の年齢は{age}です')


print_age()
print(age)
```

```python
# グローバル変数(モジュール変数)
age = 30


def print_age():
  global age
  age = 20 
  print(f'私の年齢は{age}です')

print_age()
print(age)
```

```python
# call_count = 0
CALL_COUNT = 0 # cconstant variable

def print_count1():
  global CALL_COUNT
  CALL_COUNT += 1
  print(f'関数が呼び出された回数は{CALL_COUNT}回です')

def print_count2():
  global CALL_COUNT
  CALL_COUNT += 1
  print(f'関数が呼び出された回数は{CALL_COUNT}回です')

print_count1()
print_count2()
print_count1()
print_count2()
print(CALL_COUNT)
```

## 関数の中で関数を定義する(nested function)

```python
def outer():
  def inner():
    print('This is inner function')
  inner()

outer()
```

## カプセル化：外からアクセスできないようにする

```python
def casino_entrance(age, min_age = 21):
  if age < min_age:
    print('You are too young to enter the casino')
    return

  def inner_casino_entrance():
    print('You can enter the casino')

  return inner_casino_entrance()


casino_entrance(28)
```
