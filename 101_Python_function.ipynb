## *args, **kwargs

```python
# *args タプルでもらう
def get_average(*args):
  num = len(args)
  if num == 0:
    return 0
  total = sum(args)
  return total / num

average = get_average(1, 2, 3, 4, 5)
print(average)

# **kwargs　ディクショナリーでもらう
def kwargs_func(**kwargs):
  param1 = kwargs.get('param1')
  param2 = kwargs.get('param2')
  param3 = kwargs.get('param3')
  return param1 + param2 + param3

print(kwargs_func(param1=1, param2=2, param3=3))

# *と**はunpacking, operator
numbers = (1, 2, 3)
print(*numbers)

a = {'a': 1, 'b': 2}
b = {'c': 3, 'd': 4}
print({**a, **b})
```

## 参照渡し、値渡し

```python
# 参照渡し <-> 値渡し
def add_nums(a, b):
  print(f'第一引数aのIDは{id(a)}')
  print(f'第二引数bのIDは{id(b)}')
  return a + b

one = 1
two = 2
print(f'第一引数aのIDは{id(one)}')
print(f'第二引数bのIDは{id(two)}')
print(add_nums(one, two))
```

```python
def add_one(num):
  print(f'変更前のID:{id(num)}')
  num += 1
  print(f'変更後のID:{id(num)}')
  return num


one = 1
print(id(one))
print(f'関数呼び出し前のID:{id(one)}')
add_one(one)
print(f'関数呼び出し後のID:{id(one)}')
```

## 変数のスコープ

```python
# 変数のスコープ
# ローカル変数
def print_name_local():
  first_name = 'John'
  last_name = 'Doe'
  print(f'私の名前は{first_name} {last_name}です')


print_name_local()

# グローバル変数(モジュール変数)
age = 30


def print_age():
  age = 20 # ローカル変数
  print(f'私の年齢は{age}です')


print_age()
print(age)
```

```python
# グローバル変数(モジュール変数)
age = 30


def print_age():
  global age
  age = 20 
  print(f'私の年齢は{age}です')

print_age()
print(age)
```

```python
# call_count = 0
CALL_COUNT = 0 # cconstant variable

def print_count1():
  global CALL_COUNT
  CALL_COUNT += 1
  print(f'関数が呼び出された回数は{CALL_COUNT}回です')

def print_count2():
  global CALL_COUNT
  CALL_COUNT += 1
  print(f'関数が呼び出された回数は{CALL_COUNT}回です')

print_count1()
print_count2()
print_count1()
print_count2()
print(CALL_COUNT)
```

## カプセル化：外からアクセスできないようにする

```python
def casino_entrance(age, min_age = 21):
  if age < min_age:
    print('You are too young to enter the casino')
    return

  def inner_casino_entrance():
    print('You can enter the casino')

  return inner_casino_entrance()


casino_entrance(28)
```

## クロージャ

```python
# Closure: 状態をキープした関数を作ることができる
def power(exponent):
  def inner_power(base):
    return base ** exponent
  return inner_power

power_four = power(4)
print(power_four(3))
```

##  デコレータ: 関数に機能を付属する

```python
def greeting(func):
  def inner(*args, **kwargs_func):
    print('Hello')
    func(*args, **kwargs_func)
    print('Nice to meet you')
  return inner


@ greeting
def say_name(name):
  print(F'I am {name}')

@ greeting
def say_name_and_origin(name, origin):
  print(F'I am {name} from {origin}')

# say_name('John')
 # say_name = greeting(say_name)
say_name('John')
say_name_and_origin('John', 'Japan')

```

## 再帰関数: 関数内で自信の関数をcallする関数

```python
def factrial(n):
  if n == 1:
    return 1
  else:
    return n * factrial(n - 1)

print(factrial(5))
```

### Challenge 再帰関数
Fibonacci数を再帰で実装してみる

```python
def fibonacci(num):
  if num == 0:
    return 0
  elif num == 1:
    return 1
  else:
    return fibonacci(num - 1) + fibonacci(num - 2)

print(fibonacci(6))
```

```python
def power(exponent):
  return lambda base: base ** exponent

third_power = power(3)
print(third_power(2))
```

## docstring
