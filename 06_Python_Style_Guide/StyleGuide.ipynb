# Style Guide

## 変数の定義

```python
# 変数の定義
x = 1
y = 1

# wrong
x      = 1
y      = 1
```

```python
# 関数の引数の「=」にはスペース不要
def complex(real, imag=0,0):
    return magic(r=real, i=imag)
  
# wrong
def complex(real, imag=0,0):
    return magic(r = real, i = imag)
```

```python
# operatorの前後にスペースが必要
x = 1 + 2
x = x*2 - 1
a = x*x + y*y
c = (a+1) * (a-1)

# wrong
x = 1+2
x = x * 2 - 1
a = x * x + y * y
c = (a + 1) * (a - 1)
```

```python
# カンマの後ろにはスペースを入れる
# correct
range(1, 11)
a = [1, 2, 3]

# wrong
range(1,11)
a = [1,2]

# カンマの後ろに閉じカッコがくる場合はスペース不要
# correct
foo = (0, )

# wrong
foo = (0,)
```

```python
# correct
FILES = [
  'setup.fg',
  'tox.init',
]

# wrong
FILES = ['setup.fg', 'tox.init',]
```

## 行の折り返し

```python
# correct
foo = long_function_name(a, b, c, 
                         d, e, f)

foo = long_function_name(
  a, 
  b, 
  c, 
  d, 
  e, 
  f)

# wrong
foo = long_function_name(a, b, c,
                         d, e, f)

# \で区切って改行する
# correct
a = 100000000 \ 
    + 1000000000 \
    + 100000000

# wrong
a = 100000000 + \
    1000000000 + \
    100000000


# 関数間は2行空ける
def func():
  pass


def func2():
  pass
```
